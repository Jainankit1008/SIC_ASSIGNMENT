```python
#ATM Cash Dispensing System (Greedy Algorithm)
amount = int(input("Enter withdrawal amount: ₹"))

denominations = [500, 200, 100, 50, 20]

total_notes = 0

print("\nATM Cash Dispensing Summary")
print(f"Withdrawal Amount: ₹{amount}")
print("\nDenomination Breakdown:")

for note in denominations:
    count = amount // note
    if count > 0:
        print(f"₹{note} × {count} = ₹{note * count}")
        total_notes += count
        amount %= note

print(f"\nTotal Notes Dispensed = {total_notes}")
```

```python
# University Exam Timetable Scheduler using Backtracking

subjects = [
    "Mathematics",
    "Physics",
    "Chemistry",
    "Biology",
    "Computer Science"
]

days = ["Day 1", "Day 2", "Day 3"]

conflicts = {
    "Mathematics": ["Physics", "Chemistry"],
    "Physics": ["Mathematics", "Biology"],
    "Chemistry": ["Mathematics", "Computer Science"],
    "Biology": ["Physics", "Computer Science"],
    "Computer Science": ["Chemistry", "Biology"]
}

schedule = {}

def is_safe(subject, day):
    for conflict_subject in conflicts[subject]:
        if conflict_subject in schedule and schedule[conflict_subject] == day:
            return False
    return True

def solve(index):
    if index == len(subjects):
        return True

    subject = subjects[index]

    for day in days:
        if is_safe(subject, day):
            schedule[subject] = day

            if solve(index + 1):
                return True

            del schedule[subject]

    return False

if solve(0):
    print("University Examination Schedule\n")

    for subject in subjects:
        print(f"{subject} -> {schedule[subject]}")

    print("\nSchedule Generated Successfully\n")

    print("Constraint Verification")

    checks = [
        ("Mathematics", "Physics"),
        ("Mathematics", "Chemistry"),
        ("Physics", "Biology"),
        ("Chemistry", "Computer Science"),
        ("Biology", "Computer Science")
    ]

    for s1, s2 in checks:
        if schedule[s1] != schedule[s2]:
            print(f"{s1} and {s2} -> Different Days ✓")
        else:
            print(f"{s1} and {s2} -> Same Day ✗")
else:
    print("No valid schedule found.")
```

```python
# Route Cost Optimization using Dynamic Programming

grid = [
    [1, 3, 5],
    [2, 1, 2],
    [4, 3, 1]
]

rows = len(grid)
cols = len(grid[0])

dp = [[0] * cols for _ in range(rows)]

# Fill DP table
dp[0][0] = grid[0][0]

for i in range(1, rows):
    dp[i][0] = dp[i-1][0] + grid[i][0]

for j in range(1, cols):
    dp[0][j] = dp[0][j-1] + grid[0][j]

for i in range(1, rows):
    for j in range(1, cols):
        dp[i][j] = grid[i][j] + min(dp[i-1][j], dp[i][j-1])

# Find optimal path
path = []
i, j = rows - 1, cols - 1

while i > 0 or j > 0:
    path.append((i, j))

    if i == 0:
        j -= 1
    elif j == 0:
        i -= 1
    elif dp[i-1][j] < dp[i][j-1]:
        i -= 1
    else:
        j -= 1

path.append((0, 0))
path.reverse()

# Display results
print("Transportation Cost Grid")
for row in grid:
    print(*row)

print("\nDynamic Programming Table")
for row in dp:
    print(*row)

print(f"\nMinimum Transportation Cost = {dp[rows-1][cols-1]}")

print("\nOptimal Route:")
print(" -> ".join(str(cell) for cell in path))

print("\nCost Calculation:")

costs = [grid[r][c] for r, c in path]
print(" + ".join(map(str, costs)), "=", sum(costs))
```
