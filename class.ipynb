```python
class Student:
    def __init__(self, name, marks):
        self.__name = name
        self.__marks = marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks. Marks must be between 0 and 100.")

    def get_marks(self):
        return self.__marks

    def __str__(self):
        return f"Student(Name={self.__name}, Marks={self.__marks})"


# Creating objects
s1 = Student("David", 85)
s2 = Student("John", 92)

print(s1)
print(s2)

# Updating marks
s1.set_marks(95)

print("\nAfter updating marks:")
print(s1)

# Invalid update
s1.set_marks(120)

print("Current Marks:", s1.get_marks())
```

```python
class Employee:
    def __init__(self, name, salary):
        self.__name = name
        self.__salary = salary
    def set_salary(self, salary):
        if salary > 0:
            self.__salary = salary
        else:
            print("Salary must be greater than zero.")
    def get_salary(self):
        return self.__salary
    def __str__(self):
        return f"Employee(Name={self.__name}, Salary={self.__salary})"
e1 = Employee("Alice", 50000)
e2 = Employee("Bob", 65000)
print(e1)
print(e2)
e1.set_salary(55000)
print("\nAfter salary increment:")
print(e1)
e1.set_salary(-1000)
print("Current Salary:", e1.get_salary())
```

```python
class Car:
    def __init__(self, brand, speed):
        self.__brand = brand
        self.__speed = speed
    def set_speed(self, speed):
        if speed >= 0:
            self.__speed = speed
        else:
            print("Speed cannot be negative.")
    def get_speed(self):
        return self.__speed
    def __str__(self):
        return f"Car(Brand={self.__brand}, Speed={self.__speed} km/h)"

c1 = Car("Toyota", 80)
c2 = Car("Honda", 100)
print(c1)
print(c2)
c1.set_speed(120)
print("\nAfter speed update:")
print(c1)
c1.set_speed(-20)
print(f"Current Speed: {c1.get_speed()} km/h")
```
