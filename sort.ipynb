```python
#Employee Salary Management using Merge Sort
def merge(arr, left, mid, right):
    n1 = mid - left + 1
    n2 = right - mid
    L = [0] * n1
    R = [0] * n2
    for i in range(n1):
        L[i] = arr[left + i]
    for j in range(n2):
        R[j] = arr[mid + 1 + j]
    i = 0
    j = 0
    k = left
    while i < n1 and j < n2:
        if L[i] <= R[j]:
            arr[k] = L[i]
            i += 1
        else:
            arr[k] = R[j]
            j += 1
        k += 1
    while i < n1:
        arr[k] = L[i]
        i += 1
        k += 1
    while j < n2:
        arr[k] = R[j]
        j += 1
        k += 1

def mergeSort(arr, left, right):
    if left < right:
        mid = (left + right) // 2
        mergeSort(arr, left, mid)
        mergeSort(arr, mid + 1, right)
        merge(arr, left, mid, right)

# assessment 1
n = int(input("Enter number of employees: "))
salary = []
print("Enter salaries:")
for i in range(n):
    salary.append(int(input()))
print("\nOriginal Salaries:")
for i in range(n):
    print(salary[i], end=" ")

mergeSort(salary, 0, n - 1)

print("\n\nSorted Salaries:")
for i in range(n):
    print(salary[i])
lowest = salary[0]
highest = salary[n - 1]
salary_range = highest - lowest

print("\nLowest Salary :", lowest)
print("Highest Salary:", highest)
print("Salary Range :", salary_range)


```

```python
# Flight Ticket Price Analysis using Quick Sort

def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)


n = int(input("Enter number of ticket prices: "))

prices = []

for i in range(n):
    price = int(input(f"Enter ticket price {i+1}: "))
    prices.append(price)

print("\nOriginal Ticket Prices:")
print(*prices)

sorted_prices = quick_sort(prices)

print("\nSorted Ticket Prices:")
for price in sorted_prices:
    print(price)

print("\nCheapest Ticket:", sorted_prices[0])
print("Costliest Ticket:", sorted_prices[-1])
```

```python
#Bank Transaction Processing using Merge Sort
n = int(input("Enter number of transactions: "))
transactions = []
print("Enter transaction amounts:")
for i in range(n):
    transactions.append(int(input()))
print("\nOriginal Transactions:")
for i in range(n):
    print(transactions[i], end=" ")

mergeSort(transactions, 0, n - 1)
print("\n\nSorted Transactions:")
for i in range(n):
    print(transactions[i])

minimum = transactions[0]
maximum = transactions[n - 1]

print("\nMinimum Transaction :", minimum)
print("Maximum Transaction :", maximum)
```
