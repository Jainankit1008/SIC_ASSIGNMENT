```python
books= list(range(1,10001))
book_id=int(input("Enter book ID:"))
comparison=0
def bin_search(nums,x):
    global comparison
    high=len(nums)-1
    low=0
    while (low<=high):
        comparison+=1
        mid=(low+high)//2
        if x==nums[mid]:
            return mid
        elif x>nums[mid]:
            low=mid+1
        else:
            high=mid-1
    return -1
    
if bin_search(books,book_id)==-1:
    print("Book Not Found")
else:
    print("Book Found")
    print("Position:",bin_search(books,book_id))
    print("Comparison:",comparison)
```

```python
class Stack:
    def __init__(self):
        self.items=[]
    def display(self):
        return self.items
    def isempty(self):
        if (len(self.items)==0):
            return "The stack is empty"
    def push(self,items):
        self.items.append(items)
    def pop(self):
        if self.isempty():
            return "the stack is empty"
        else:
            return self.items.pop()
    def peek(self):
        if self.isempty():
            return "the stack is empty"
        else:
            return self.items[-1]
railparcel = Stack()
while True:
    print("\n1.Load 2.Unload 3.Peek 4.Display 5.Exit")
    ch = int(input("Choice: "))
    if ch == 1:
        railparcel.push(input("Parcel ID: "))
    elif ch == 2:
        item = railparcel.pop()
        if item is None:
            print("No Parcels Available For Unloading")
        else:
            print("Removed Parcel:", item)
    elif ch == 3:
        item = railparcel.peek()
        if item is None:
            print("No Parcels Available")
        else:
            print("Next Parcel:", item)
    elif ch == 4:
        if railparcel.isempty():
            print("Wagon Empty")
        else:
            print("Parcels:", railparcel.display())
    elif ch == 5:
        break
    else:
        print("Invalid Choice")
```

```python
from collections import deque

class Queue:
    def __init__(self):
        self.queue = deque()

    def enqueue(self, passenger):
        self.queue.append(passenger)

    def dequeue(self):
        if not self.is_empty():
            return self.queue.popleft()
        return None

    def front(self):
        if not self.is_empty():
            return self.queue[0]
        return None

    def is_empty(self):
        return len(self.queue) == 0

    def display(self):
        if self.is_empty():
            print("No Passengers Waiting")
        else:
            print("Passengers Waiting:")
            for passenger in self.queue:
                print(passenger)


reservation = Queue()

while True:
    print("\n1. Add Passenger")
    print("2. Serve Passenger")
    print("3. View Next Passenger")
    print("4. Display Queue")
    print("5. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        name = input("Enter Passenger Name: ")
        reservation.enqueue(name)
        print("Passenger Added Successfully")

    elif choice == 2:
        served = reservation.dequeue()

        if served:
            print("Passenger Served:")
            print(served)
        else:
            print("No Passengers In Queue")

    elif choice == 3:
        next_passenger = reservation.front()

        if next_passenger:
            print("Next Passenger:")
            print(next_passenger)
        else:
            print("Queue Empty")

    elif choice == 4:
        reservation.display()

    elif choice == 5:
        print("Program Ended")
        break

    else:
        print("Invalid Choice")
```
